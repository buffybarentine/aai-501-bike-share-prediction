```python
# Import libraries
import pandas as pd
from pathlib import Path
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
# import data
data_df = pd.read_csv("data/raw/hour.csv")

data_df.head(3)
```

```python
# Assign date column as datetime type and create datetime column
data_df["dteday"] = pd.to_datetime(data_df["dteday"])
data_df["datetime"] = (
    data_df["dteday"] + pd.to_timedelta(data_df["hr"], unit="h")
)

# Sort by datetime and reset index
data_df = data_df.sort_values("datetime", ignore_index=True)

data_df.head(3)
```

```python
# view column types
data_df.dtypes
```

```python
# Remove columns 
drop_cols = ["instant", "casual", "registered", "atemp"]
processed_df = data_df.drop(columns=drop_cols)
processed_df.head(3)
```

```python
# Mark categorical columns as category type
categorical_cols = [
    "season",
    "mnth",
    "hr",
    "weekday",
    "weathersit"
]

processed_df[categorical_cols] = processed_df[categorical_cols].astype("category")

processed_df.info()
```

```python
# Export as parquet file to retain categorical types
processed_df.to_parquet(
    "data/processed/boosted_tree.parquet",
    engine="pyarrow",
    index=False
)
print("Processed data exported as parquet file.")
```
